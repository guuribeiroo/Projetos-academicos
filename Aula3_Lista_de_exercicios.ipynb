```python
print("Calculando Área do triângulo")

altura = float(input("Insira a medida da altura:\n"))
base = float(input("Insira a medida da base: \n"))

area = (base * altura) /2
print("A medida da área do triangulo é :", area)
```

```python
print("Calculando o perimetro do triangulo")

l1 = float(input("Insira a medida do lado 1: \n"))
l2 = float(input("Insira a medida do lado 2 : \n"))

perimetro = (l1 * 2) + (l2*2)

print("O perimetro total do retangulo é: ",perimetro)
```

```python
print("Identificando número par ou impar")

numero = int(input("Insira um número inteiro qualquer: \n"))

if numero % 2 == 0:
  print("Esse número é PAR")
else:
  print("Esse número é IMPAR")


# % é o resto da divisão efetuada
# == é o igual a
```

```python
print("Identificando negativo, positivo ou zero")

numero = int(input("Insira qualquer número inteiro: \n"))

if numero > 0:
  print("Esse número é POSITIVO")
elif numero < 0:
  print("Esse número é NEGATIVO")
else:
  print("É 0")
```
