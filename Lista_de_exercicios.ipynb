```python
print("Identificando número par ou impar")

numero = int(input("Insira um número inteiro qualquer: \n"))

if numero % 2 == 0:
  print("Esse número é PAR")
else:
  print("Esse número é IMPAR")


# % é o resto da divisão efetuada
# == é o igual a
```

```python
print("Identificando negativo, positivo ou zero")

numero = int(input("Insira qualquer número inteiro: \n"))

if numero > 0:
  print("Esse número é POSITIVO")
elif numero < 0:
  print("Esse número é NEGATIVO")
else:
  print("É 0")
```

```python
print("Identificando o maior entre 3 números")
a = int(input("Insira o primeiro número: \n"))
b = int(input("Insira o segundo número: \n"))
c = int(input("Insira o terceiro número: \n"))

if a > b and a > c:
  print(a,"É o maior número entre os 3")
elif b > c and b > a:
  print(b,"É o maior número entre os 3")
else:
  print(c,"É o maior número entre os 3")
```

```python

print("Quem pode votar?")

idade = int(input("Idade: \n"))

titulo_eleitor = input("Possui título de eleitor? \n")

if titulo_eleitor.lower() == "sim":
  titulo_eleitor = True
else:
  titulo_eleitor = False

if idade >= 16 and titulo_eleitor:
  print("Você pode votar!")

else:
  print("Você não pode votar")


```
